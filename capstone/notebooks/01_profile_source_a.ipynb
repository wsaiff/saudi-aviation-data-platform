# Source A profiling

Profile the raw Source A extract to establish its grain, candidate key, null patterns, coverage, and any surprises before defining a production contract. No schema or results are assumed here.

## Inputs and assumptions

Place a local extract at `data/source_a.csv` or update `SOURCE_PATH`. Confirm the extract's time window, timezone, source-system meaning, and intended row grain with the source owner before treating any finding as a contract.

```python
from pathlib import Path

import pandas as pd

SOURCE_PATH = Path("data/source_a.csv")
df = pd.read_csv(SOURCE_PATH)

print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns):,}")
display(df.head())
```

```python
profile = pd.DataFrame({
    "dtype": df.dtypes.astype(str),
    "null_count": df.isna().sum(),
    "null_rate": df.isna().mean(),
    "distinct_count": df.nunique(dropna=True),
}).sort_values("null_rate", ascending=False)
display(profile)
```

```python
duplicate_rows = int(df.duplicated().sum())
print(f"Full-row duplicates: {duplicate_rows:,}")
```

```python
# Populate only after a source owner confirms the business key.
CANDIDATE_KEY: list[str] = []

if not CANDIDATE_KEY:
    print("Candidate key not configured; confirm the grain and key with the source owner.")
else:
    missing_key_columns = sorted(set(CANDIDATE_KEY) - set(df.columns))
    if missing_key_columns:
        raise KeyError(f"Candidate key columns not found: {missing_key_columns}")
    key_rows = df[CANDIDATE_KEY]
    print(f"Key columns: {CANDIDATE_KEY}")
    print(f"Rows with null key values: {int(key_rows.isna().any(axis=1).sum()):,}")
    print(f"Duplicate key rows: {int(key_rows.duplicated().sum()):,}")
```
